SQL_Task



*   **SQL Definition**: Structured Query Language is the standard language for managing and manipulating relational databases.
*   **SQL Use**: Used to store, retrieve, manage, and organize data efficiently within relational databases.
*   **Real-world Usage**: Powers data management in applications like social media platforms, e-commerce sites, and banking systems.
*   **DDL (Data Definition Language)**: Commands for defining and modifying the database structure (e.g., `CREATE`, `ALTER`, `DROP`).
*   **DML (Data Manipulation Language)**: Commands for managing data within database objects (e.g., `SELECT`, `INSERT`, `UPDATE`, `DELETE`).
*   **DCL (Data Control Language)**: Commands for controlling database access and permissions (e.g., `GRANT`, `REVOKE`).
*   **TCL (Transaction Control Language)**: Commands for managing database transactions to ensure data integrity (e.g., `COMMIT`, `ROLLBACK`).

## SQL Data Types Notes

*   **SQL Data Types Definition**: Data types classify the kind of values that can be stored in a column, defining the range and type of data.
*   **INT**: Used for storing whole numbers (integers).
*   **VARCHAR**: Used for storing variable-length character strings (text).
*   **DATE**: Used for storing date values (year, month, day).
*   **FLOAT**: Used for storing approximate floating-point numbers (decimal numbers).
*   **BOOLEAN**: Used for storing true/false or 0/1 values.


**CREATE TABLE**: Used to create a new table in the database with specified columns and data types.

**INSERT INTO**: Used to add records (rows) into an existing table.

**FOR EXAMPLE**:Student Table

```python
import sqlite3
from IPython.display import display
import pandas as pd

# Create SQLite connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create Students table
cursor.execute("""
CREATE TABLE Students (
    ID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    Gender TEXT,
    Department TEXT,
    Marks INTEGER,
    City TEXT
)
""")

# Insert sample records
cursor.executemany("""
INSERT INTO Students VALUES (?, ?, ?, ?, ?, ?, ?)
""", [
    (1, 'Alice', 20, 'Female', 'CSE', 85, 'Chennai'),
    (2, 'Bob', 21, 'Male', 'ECE', 78, 'Bangalore'),
    (3, 'Charlie', 19, 'Male', 'IT', 88, 'Hyderabad'),
    (4, 'Diana', 22, 'Female', 'CSE', 92, 'Coimbatore'),
    (5, 'Eve', 20, 'Female', 'EEE', 74, 'Madurai')
])

# Fetch and display table
df = pd.read_sql_query("SELECT * FROM Students", conn)
display(df)
```

**SELECT with WHERE Conditions**

**SELECT**: Used to retrieve specific columns or records from a table.

**WHERE**: Used to filter records based on conditions.

**AND**: Used to combine multiple conditions in a query.

```python
# SELECT specific columns from the Students table
query1 = "SELECT Name, Department, Marks FROM Students"
df1 = pd.read_sql_query(query1, conn)
display(df1)

# SELECT students who scored more than 80 marks using WHERE clause
query2 = "SELECT * FROM Students WHERE Marks > 80"
df2 = pd.read_sql_query(query2, conn)
display(df2)

# SELECT students from CSE department using WHERE condition
query3 = "SELECT * FROM Students WHERE Department = 'CSE'"
df3 = pd.read_sql_query(query3, conn)
display(df3)

# SELECT students with marks greater than 80 AND age less than 22
query4 = "SELECT * FROM Students WHERE Marks > 80 AND Age < 22"
df4 = pd.read_sql_query(query4, conn)
display(df4)
```

**Aggregate Functions:**

**COUNT()**: Returns the total number of records.

**SUM():** Returns the total sum of a numeric column.

**AVG():** Returns the average value of a numeric column.

**MIN():** Returns the minimum value in a column.

**MAX():** Returns the maximum value in a column.

```python
# COUNT the total number of students
query1 = "SELECT COUNT(*) AS Total_Students FROM Students"
df1 = pd.read_sql_query(query1, conn)
display(df1)

# SUM of marks of all students
query2 = "SELECT SUM(Marks) AS Total_Marks FROM Students"
df2 = pd.read_sql_query(query2, conn)
display(df2)

# AVERAGE marks of students
query3 = "SELECT AVG(Marks) AS Average_Marks FROM Students"
df3 = pd.read_sql_query(query3, conn)
display(df3)

# MINIMUM marks obtained
query4 = "SELECT MIN(Marks) AS Minimum_Marks FROM Students"
df4 = pd.read_sql_query(query4, conn)
display(df4)

# MAXIMUM marks obtained
query5 = "SELECT MAX(Marks) AS Maximum_Marks FROM Students"
df5 = pd.read_sql_query(query5, conn)
display(df5)
```

**GROUP BY:** Groups records that have the same values in a column.

**HAVING:** Filters grouped records based on aggregate conditions.

**Difference:** WHERE filters rows, HAVING filters groups.

```python
 # GROUP students by Department and calculate average marks
query1 = """
SELECT Department, AVG(Marks) AS Average_Marks
FROM Students
GROUP BY Department
"""
df1 = pd.read_sql_query(query1, conn)
display(df1)

# GROUP BY with HAVING to show departments with average marks above 80
query2 = """
SELECT Department, AVG(Marks) AS Average_Marks
FROM Students
GROUP BY Department
HAVING AVG(Marks) > 80
"""
df2 = pd.read_sql_query(query2, conn)
display(df2)
```

**ORDER BY: **Sorts the result set in ascending or descending order.

**ASC:** Sorts data in ascending order (default).

**DESC:** Sorts data in descending order.

**LIMIT:** Restricts the number of records displayed.

```python
# ORDER students by Marks in ascending order
query1 = "SELECT * FROM Students ORDER BY Marks ASC"
df1 = pd.read_sql_query(query1, conn)
display(df1)

# ORDER students by Marks in descending order
query2 = "SELECT * FROM Students ORDER BY Marks DESC"
df2 = pd.read_sql_query(query2, conn)
display(df2)

# Display only top 3 students based on highest marks
query3 = "SELECT * FROM Students ORDER BY Marks DESC LIMIT 3"
df3 = pd.read_sql_query(query3, conn)
display(df3)
```

**JOIN:** Used to combine rows from two tables based on a related column.

**INNER JOIN:** Returns matching records from both tables.

**LEFT JOIN:** Returns all records from the left table and matching records from the right table.

```python
# Create Departments table
cursor.execute("""
CREATE TABLE Departments (
    Department TEXT PRIMARY KEY,
    HOD TEXT
)
""")

# Insert records into Departments table
cursor.executemany("""
INSERT INTO Departments VALUES (?, ?)
""", [
    ('CSE', 'Dr. Kumar'),
    ('ECE', 'Dr. Rao'),
    ('IT', 'Dr. Meena'),
    ('EEE', 'Dr. Sharma')
])

# INNER JOIN between Students and Departments
query1 = """
SELECT Students.Name, Students.Department, Departments.HOD
FROM Students
INNER JOIN Departments
ON Students.Department = Departments.Department
"""
df1 = pd.read_sql_query(query1, conn)
display(df1)

# LEFT JOIN between Students and Departments
query2 = """
SELECT Students.Name, Students.Department, Departments.HOD
FROM Students
LEFT JOIN Departments
ON Students.Department = Departments.Department
"""
df2 = pd.read_sql_query(query2, conn)
display(df2)
```

**SELF JOIN:** A join where a table is joined with itself.

**OUTER JOIN:** Returns matching records and also non-matching records from one or both tables.

**LEFT OUTER JOIN:** Returns all records from the left table and matching records from the right table.

**RIGHT OUTER JOIN:** Returns all records from the right table and matching records from the left table.

```python
# SELF JOIN on Students table to compare students from the same department
query1 = """
SELECT A.Name AS Student1, B.Name AS Student2, A.Department
FROM Students A
JOIN Students B
ON A.Department = B.Department
AND A.ID <> B.ID
"""
df1 = pd.read_sql_query(query1, conn)
display(df1)

# LEFT OUTER JOIN between Students and Departments
query2 = """
SELECT Students.Name, Students.Department, Departments.HOD
FROM Students
LEFT OUTER JOIN Departments
ON Students.Department = Departments.Department
"""
df2 = pd.read_sql_query(query2, conn)
display(df2)
```

**Subquery:** A query written inside another SQL query.

**Purpose:** Used to filter or compare values based on another query result.

```python
# Select students who scored more than the average marks
query1 = """
SELECT *
FROM Students
WHERE Marks > (
    SELECT AVG(Marks)
    FROM Students
)
"""
df1 = pd.read_sql_query(query1, conn)
display(df1)
```

**VIEW:** A virtual table created using a SELECT query.

```python
# Create a view for students with marks above 80
cursor.execute("""
CREATE VIEW HighScorers AS
SELECT Name, Department, Marks
FROM Students
WHERE Marks > 80
""")

# Retrieve data from the view
df = pd.read_sql_query("SELECT * FROM HighScorers", conn)
display(df)
```

**TRANSACTION:** A set of SQL operations executed as a single unit

```python
# Update marks of a student (transaction starts automatically)
cursor.execute("UPDATE Students SET Marks = 90 WHERE ID = 2")

# Rollback the transaction
conn.rollback()

# Display table after rollback
df = pd.read_sql_query("SELECT * FROM Students", conn)
display(df)
```
